Step 2.1: Track A - Replicating The DID Engine

```python
# Construct FTE employment (Paper p.775: full-time + managers + 0.5 * part-time)
df_raw['FTE'] = df_raw['EMPFT'] + df_raw['NMGRS'] + 0.5 * df_raw['EMPPT']
df_raw['FTE2'] = df_raw['EMPFT2'] + df_raw['NMGRS2'] + 0.5 * df_raw['EMPPT2']
```

```python
df_raw[['SHEET', 'STATE', 'FTE', 'FTE2', 'WAGE_ST', 'WAGE_ST2']].head()
```

Task 1: Descriptive Table Reconstruction (Table 2)

```python
nj = df_raw[df_raw['STATE'] == 1]
pa = df_raw[df_raw['STATE'] == 0]

print('Wave 1')
print('NJ FTE mean:', nj['FTE'].mean())
print('PA FTE mean:', pa['FTE'].mean())
print('NJ starting wage mean:', nj['WAGE_ST'].mean())
print('PA starting wage mean:', pa['WAGE_ST'].mean())

print('\nWave 2')
print('NJ FTE mean:', nj['FTE2'].mean())
print('PA FTE mean:', pa['FTE2'].mean())
print('NJ starting wage mean:', nj['WAGE_ST2'].mean())
print('PA starting wage mean:', pa['WAGE_ST2'].mean())
```

```python
print('Standard Deviations')
print('Wave 1')
print('NJ FTE std:', nj['FTE'].std())
print('PA FTE std:', pa['FTE'].std())
print('NJ starting wage std:', nj['WAGE_ST'].std())
print('PA starting wage std:', pa['WAGE_ST'].std())

print('\nWave 2')
print('NJ FTE std:', nj['FTE2'].std())
print('PA FTE std:', pa['FTE2'].std())
print('NJ starting wage 2 std:', nj['WAGE_ST2'].std())
print('PA starting wage 2 std:', pa['WAGE_ST2'].std())
```

Task 2: The Simple Difference (DID by hand)

```python
# Calculate mean FTE change for NJ and PA
nj_diff = nj['FTE2'].mean() - nj['FTE'].mean()
pa_diff = pa['FTE2'].mean() - pa['FTE'].mean()
did = nj_diff - pa_diff

print('NJ FTE change:', nj_diff)
print('PA FTE change:', pa_diff)
print('DID estimate:', did)
```

Task 3: Regression Implementatoin (DID Regression)

```python
import statsmodels.formula.api as smf

# Create long format dataset
wave1 = df_raw[['SHEET', 'STATE', 'FTE']].copy()
wave1['POST'] = 0

wave2 = df_raw[['SHEET', 'STATE', 'FTE2']].copy()
wave2['POST'] = 1
wave2 = wave2.rename(columns={'FTE2': 'FTE'})

df_long = pd.concat([wave1, wave2])

# DID regression
model = smf.ols('FTE ~ STATE * POST', data=df_long).fit()
print(model.summary())
```

Task 4: Standard Errors and Clustering

```python
df_long_clean = df_long.dropna(subset=['FTE'])

# DID regression with clustered standard errors at store level
results = smf.ols('FTE ~ STATE * POST', data=df_long_clean).fit(cov_type='cluster', cov_kwds={'groups': df_long_clean['SHEET']})
print(results.summary())
```
